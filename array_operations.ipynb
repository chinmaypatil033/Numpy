```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
```

```python
arr=np.vstack((a,b))
arr
```

```python
arr=np.hstack((a,b))
arr
```

```python
arr=np.column_stack((a,b))
arr
```

```python
arr=np.row_stack((a,b))
arr
```
