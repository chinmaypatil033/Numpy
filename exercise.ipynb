```python
c=np.sum(arr,axis=1)
c[c==45]
```

```python
# columns=[age,maths marks, science marks]
import numpy as np
data=np.array([[18,85,78],
               [19,95,80],
               [20,80,90],
               [21,90,85]])
```

```python
#  SHAPE OF MATRIX
data.shape
```

```python
# Mean of age column 
f=data[:,:1]
f.mean()
```

```python
# Extract marks of maths
g=data[:,1:2]
g
```

```python
#  highest marks in science
h=max(data[:,2:3])
h
```

```python
# student got marks greater than 90 in maths 
data[data[:,1]>90]
```

```python
#  INCREASE MATHS MARKS OF ALL STUDENTS BY 5
data[:,1]+=5
data
```

```python
#  STUDENTS YOUNGER THAN 19
len(data[data[:,0]<19])
```

```python
#  CALCULATE AVERAGE OF EACH SUBJECT
np.mean(data[:,1:],axis=0)
```
